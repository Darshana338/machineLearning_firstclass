```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("darshana.csv")
```

```python
data
```

```python
concept = np.array(data)[:,:-1]
```

```python
concept
```

```python
target = np.array(data)[:,-1]
```

```python
target
```

```python
def train(con, tar):
  
    for i, val in enumerate(tar):
        if val == "yes":
            specific_h = con[i].copy()
            break
            
  
    for i, val in enumerate(con):
        if tar[i] == "yes":
            for x in range(len(specific_h)):
                if val[x] != specific_h[x]:
                    specific_h[x] = "?"
                else:
                    pass
    
   
    return specific_h
```

```python
print(train(concept,target))
```
